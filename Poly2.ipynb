```python
# ===============================
# 1) Imports
# ===============================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
    classification_report, confusion_matrix, roc_curve
)

# ===============================
# 2) Train-test split
# ===============================
X1 = df1.drop('Diabetes_012', axis=1)
y1 = df1['Diabetes_012']

# Keep only binary outcome (0 = non-diabetic, 2 = diabetic)
mask = (y1 != 1)  # remove pre-diabetes
X1 = X1[mask]
y1 = y1[mask].replace(2, 1)  # map 2 → 1

X1_train, X1_test, y1_train, y1_test = train_test_split(
    X1, y1, test_size=0.3, random_state=42, stratify=y1
)

# ===============================
# 3) Define models
# ===============================
L1 = LogisticRegression(penalty='l1', solver='liblinear', random_state=42)
model3 = LogisticRegression(penalty='l2', solver='liblinear', random_state=42)
clf1 = RandomForestClassifier(n_estimators=100, random_state=42)
dummy_clf = DummyClassifier(strategy="most_frequent")

models = {
    "LogReg L1": L1,
    "LogReg L2": model3,
    "Random Forest": clf1,
    "Dummy Classifier": dummy_clf
}

# ===============================
# 4) Train + Predict + Evaluate
# ===============================
def show_metrics(name, y_true, y_pred, y_prob=None):
    print(f"\n=== {name} ===")
    print("Accuracy :", accuracy_score(y_true, y_pred))
    print("Precision:", precision_score(y_true, y_pred, zero_division=0))
    print("Recall   :", recall_score(y_true, y_pred, zero_division=0))
    print("F1-score :", f1_score(y_true, y_pred, zero_division=0))
    if y_prob is not None:
        print("ROC-AUC  :", roc_auc_score(y_true, y_prob))
    print(classification_report(y_true, y_pred, target_names=['non-diabetic','diabetic']))

    # Confusion matrix
    cm = confusion_matrix(y_true, y_pred)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=['Non-diabetic','Diabetic'], yticklabels=['Non-diabetic','Diabetic'])
    plt.title(f"{name} Confusion Matrix")
    plt.ylabel("True")
    plt.xlabel("Predicted")
    plt.show()

# Train + evaluate each
results = {}
for name, model in models.items():
    model.fit(X1_train, y1_train)
    y_pred = model.predict(X1_test)
    y_prob = model.predict_proba(X1_test)[:,1] if hasattr(model,"predict_proba") else None
    show_metrics(name, y1_test, y_pred, y_prob)
    results[name] = (y_pred, y_prob)

# ===============================
# 5) ROC Curves
# ===============================
plt.figure(figsize=(7,6))

for name, (y_pred, y_prob) in results.items():
    if y_prob is not None:
        fpr, tpr, _ = roc_curve(y1_test, y_prob)
        auc = roc_auc_score(y1_test, y_prob)
        plt.plot(fpr, tpr, label=f"{name} (AUC={auc:.3f})")

plt.plot([0,1],[0,1],'k--',label="Chance")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curves")
plt.legend()
plt.show()

```
